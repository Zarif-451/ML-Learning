### Confidence Interval in a Bar Plot

In a Seaborn **Bar Plot**, the height of each bar represents the **mean (average)** of the numerical variable for each category.

The **black vertical line** on top of each bar is the **Confidence Interval (CI)**, which represents the uncertainty around the estimated mean.

For example, in the plot below,

```python
sns.barplot(data=titanic, x="Pclass", y="Age")
```

- The height of each bar is

$$
\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i,
$$

where

- $\bar{x}$ = Sample mean
- $n$ = Number of observations
- $x_i$ = Individual observation

The confidence interval is calculated as

$$
\text{CI}=\bar{x}\pm t\left(\frac{s}{\sqrt{n}}\right),
$$

where

- $\bar{x}$ = Sample mean
- $t$ = Critical value from the $t$-distribution
- $s$ = Sample standard deviation
- $n$ = Sample size

### Interpretation

- **Bar Height** → Mean (Average)
- **Black Error Bar** → Confidence Interval (CI)

A **longer** confidence interval indicates **greater uncertainty** in the estimated mean.

A **shorter** confidence interval indicates **less uncertainty** and a more precise estimate of the mean.

> **Note:** By default, Seaborn displays a **95% Confidence Interval** for the estimated mean.

## Choosing the Right Pandas Function Based on Variable Types

| Variable Types | Common Pandas Function | Common Visualization |
|----------------|------------------------|----------------------|
| **Categorical** | `value_counts()` | Count Plot (`sns.countplot()`) |
| **Categorical + Categorical** | `pd.crosstab()` | Heatmap (`sns.heatmap()`) |
| **Categorical + Numerical** | `groupby()` | Bar Plot, Box Plot, Violin Plot |
| **Numerical + Numerical** | `corr()`, direct column selection | Scatter Plot, Correlation Heatmap |

### Rule of Thumb

- **One Categorical Variable** → `value_counts()`
- **Two Categorical Variables** → `pd.crosstab()`
- **Categorical + Numerical** → `groupby()`
- **Two Numerical Variables** → `corr()`, `scatterplot()`

### Binary Variables in Machine Learning

In Machine Learning, binary variables such as:

- **Survived** (0/1)
- **Purchased** (0/1)
- **Spam** (0/1)
- **Churn** (0/1)
- **Fraud** (0/1)

are **categorical in meaning**, since they represent distinct classes.

However, because they are encoded as **0** and **1**, many Machine Learning algorithms and statistical operations treat them as numerical values.

For example,

$$
\text{Mean} = \frac{\sum_{i=1}^{n} x_i}{n}
$$

where

- $x_i \in \{0,1\}$
- $n$ = Number of observations

Since the values are only **0** and **1**, the mean becomes

$$
\text{Mean}=\frac{\text{Number of 1's}}{\text{Total Number of Observations}}
$$

which represents the **proportion (or percentage) of positive cases**.

For example,

$$
\frac{1+0+1+1+0}{5}
=
\frac{3}{5}
=
0.6
=
60\%
$$

Thus, although binary variables are **categorical**, operations such as `mean()` can be used to calculate proportions or percentages.

# Extras

# `value_counts()` vs `pd.crosstab()`

Both functions are used with **categorical variables**, but they answer different questions.

---

## `value_counts()`

Use `value_counts()` when working with **one categorical variable**.

It answers the question:

> **"How many observations belong to each category?"**

### Example

```python
titanic["Sex"].value_counts()
```

Output:

| Sex | Count |
|------|------:|
| Male | 577 |
| Female | 314 |

**Interpretation:**

- Number of male passengers.
- Number of female passengers.

---

## `pd.crosstab()`

Use `pd.crosstab()` when working with **two categorical variables**.

It answers the question:

> **"How many observations belong to each combination of categories?"**

### Example

```python
pd.crosstab(
    titanic["Sex"],
    titanic["Survived"]
)
```

Output:

| Sex | Died (0) | Survived (1) |
|------|---------:|-------------:|
| Female | 81 | 233 |
| Male | 468 | 109 |

**Interpretation:**

- Number of females who survived and did not survive.
- Number of males who survived and did not survive.

---

## Comparison

| Function | Number of Categorical Variables | Answers |
|----------|---------------------------------|----------|
| `value_counts()` | **One** | "How many of each category?" |
| `pd.crosstab()` | **Two** | "How many in each combination of categories?" |

---

## Examples

### One Categorical Variable

```python
titanic["Pclass"].value_counts()
```

Question:

> **How many passengers are in each passenger class?**

---

### Two Categorical Variables

```python
pd.crosstab(
    titanic["Pclass"],
    titanic["Survived"]
)
```

Question:

> **How many First-Class passengers survived?**

> **How many Third-Class passengers did not survive?**

---

## Rule of Thumb

- **One categorical variable** → `value_counts()`
- **Two categorical variables** → `pd.crosstab()`

> **Easy way to remember:**  
> Ask yourself, **"How many categorical variables am I analyzing?"**
>
> - **1 Variable** → `value_counts()`
> - **2 Variables** → `pd.crosstab()`

# Extras

# `groupby()`

`groupby()` is used when you have **one categorical variable** and **one numerical variable**.

It answers questions such as:

- What is the **average**?
- What is the **maximum**?
- What is the **minimum**?
- What is the **median**?
- What is the **standard deviation?**

for each category.

---

## Why do we use `groupby()`?

Unlike `value_counts()` and `pd.crosstab()`, which **count observations**, `groupby()` performs **statistical calculations** on numerical data after grouping by a categorical variable.

---

## General Syntax

```python
df.groupby("Categorical_Column")["Numerical_Column"].aggregation()
```

where `aggregation()` can be:

- `mean()`
- `max()`
- `min()`
- `median()`
- `sum()`
- `std()`
- `count()`

---

## Example 1: Mean

```python
titanic.groupby("Pclass")["Fare"].mean()
```

Question:

> **What is the average fare paid by passengers in each passenger class?**

Output:

| Pclass | Mean Fare |
|--------:|----------:|
| 1 | 84.15 |
| 2 | 20.66 |
| 3 | 13.67 |

---

## Example 2: Maximum

```python
titanic.groupby("Pclass")["Fare"].max()
```

Question:

> **What is the highest fare paid in each passenger class?**

---

## Example 3: Minimum

```python
titanic.groupby("Pclass")["Fare"].min()
```

Question:

> **What is the lowest fare paid in each passenger class?**

---

## Example 4: Median

```python
titanic.groupby("Pclass")["Fare"].median()
```

Question:

> **What is the median fare for each passenger class?**

---

## Example 5: Binary Variable

```python
titanic.groupby("Sex")["Survived"].mean()
```

Output:

| Sex | Survival Rate |
|------|--------------:|
| Female | 0.742 |
| Male | 0.189 |

Although **`Survived` is categorical**, it is encoded as **0** and **1**.

Therefore,

$$
\text{Mean}=\frac{\text{Number of Survivors}}{\text{Total Number of Passengers}}
$$

The mean represents the **proportion (or percentage) of passengers who survived**.

---

## Comparison

| Function | Variable Types | Purpose |
|----------|----------------|---------|
| `value_counts()` | One Categorical | Count each category |
| `pd.crosstab()` | Two Categorical | Count combinations of categories |
| `groupby()` | Categorical + Numerical | Calculate statistics for each category |

---

# Choosing the X-axis and Y-axis

Selecting the correct variables for the **X-axis** and **Y-axis** depends on the **types of variables** being analyzed.

---

## 1. Categorical + Numerical

For a **categorical** and a **numerical** variable:

- **X-axis** → Categorical Variable
- **Y-axis** → Numerical Variable

This allows the numerical values to be compared across different categories.

### Example

```python
sns.barplot(
    data=titanic,
    x="Pclass",
    y="Fare"
)
```

**Interpretation:**

- **X-axis:** Passenger Class (Category)
- **Y-axis:** Fare (Numerical)

---

## 2. Numerical + Numerical

For two **numerical** variables:

- **X-axis** → Independent Variable
- **Y-axis** → Dependent Variable

This is commonly used with **Scatter Plots** and **Line Plots**.

### Example

```python
sns.scatterplot(
    data=titanic,
    x="Age",
    y="Fare"
)
```

**Interpretation:**

- **X-axis:** Age
- **Y-axis:** Fare

---

## 3. Categorical + Categorical

When both variables are **categorical**, there is usually **no numerical Y-axis**.

Instead, use:

- `pd.crosstab()` + `sns.heatmap()`, or
- `sns.countplot()` with `hue`.

### Example

```python
pd.crosstab(
    titanic["Sex"],
    titanic["Survived"]
)
```

or

```python
sns.countplot(
    data=titanic,
    x="Sex",
    hue="Survived"
)
```

---

## Summary

| Variable Types | X-axis | Y-axis | Common Plot |
|----------------|--------|--------|-------------|
| **Categorical + Numerical** | Categorical | Numerical | Bar Plot, Box Plot, Violin Plot |
| **Numerical + Numerical** | Independent Variable | Dependent Variable | Scatter Plot, Line Plot |
| **Categorical + Categorical** | Category | Count (Automatic) or Second Category (`hue`) | Count Plot, Heatmap |

---

## Rule of Thumb

1. **Categorical + Numerical**
   - **X-axis** → Categorical
   - **Y-axis** → Numerical

2. **Numerical + Numerical**
   - **X-axis** → Independent Variable
   - **Y-axis** → Dependent Variable

3. **Categorical + Categorical**
   - Use **`pd.crosstab()` + `sns.heatmap()`**
   - Or use **`sns.countplot(hue=...)`**

> **Easy way to remember:**
>
> - **Category → X-axis**
> - **Number → Y-axis**
> - **Two Numbers → Scatter Plot**
> - **Two Categories → Crosstab + Heatmap (or Count Plot with `hue`)**

# `corr()`

`corr()` is used to measure the **relationship (correlation)** between **two or more numerical variables**.

It answers questions such as:

- Does one variable increase when another increases?
- Does one variable decrease when another increases?
- Is there any relationship between the variables?

---

## Why do we use `corr()`?

Unlike `groupby()` or `pd.crosstab()`, which summarize data by categories, `corr()` measures the **strength and direction of a linear relationship** between numerical variables.

---

## General Syntax

```python
df.corr(numeric_only=True)
```

or

```python
df[["Column1", "Column2"]].corr()
```

---

## Example

```python
titanic[["Age", "Fare"]].corr()
```

Output

| | Age | Fare |
|------|------:|------:|
| Age | 1.00 | 0.10 |
| Fare | 0.10 | 1.00 |

---

## Interpretation

The correlation coefficient ranges from

$$
-1 \le r \le 1
$$

where

| Correlation ($r$) | Interpretation |
|------------------:|---------------|
| **+1** | Perfect Positive Correlation |
| **0** | No Linear Correlation |
| **-1** | Perfect Negative Correlation |

---

## Types of Correlation

### Positive Correlation

As one variable increases, the other also increases.

Example:

- Study Hours ↑
- Exam Marks ↑

$$
r > 0
$$

---

### Negative Correlation

As one variable increases, the other decreases.

Example:

- Speed ↑
- Travel Time ↓

$$
r < 0
$$

---

### No Correlation

There is no clear linear relationship.

$$
r \approx 0
$$

---

## Visualizing Correlation

The most common visualizations are:

### Scatter Plot

```python
sns.scatterplot(
    data=titanic,
    x="Age",
    y="Fare"
)
```

---

### Correlation Heatmap

```python
sns.heatmap(
    titanic.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)
```

The heatmap colors make it easier to identify strong positive and negative correlations.

---
