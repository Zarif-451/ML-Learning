```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head(5)
```

```python
pd.crosstab(tips["sex"], tips["smoker"])
```

```python
sns.heatmap(pd.crosstab(tips["sex"], tips["smoker"]), annot=True)
```

```python
sns.countplot(data=tips, x="sex", hue="smoker")
```

```python
tips.groupby("smoker", observed=True)["total_bill"].sum()
```

```python
tips.groupby("smoker", observed=True).mean(numeric_only=True)
```

```python
tips.groupby("smoker", observed=True).std(numeric_only=True)
```

```python
tips.groupby("smoker", observed=True).median(numeric_only=True)
```

```python
tips.groupby("sex", observed=True).agg({
    "total_bill" : ["mean", "sum", "max", "min"],
    "tip" : ["max", "min"]
})
```

```python
tips.groupby("smoker", observed=True).describe()
```
