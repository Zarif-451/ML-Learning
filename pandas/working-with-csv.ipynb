# 10. Encoding Parameter

# Common CSV Encodings

| Encoding | Where it's commonly used | Example in Pandas |
|----------|--------------------------|-------------------|
| **utf-8** ⭐ | Modern standard (GitHub, Kaggle, APIs, Linux, macOS) | `encoding="utf-8"` *(default)* |
| **latin-1** (ISO-8859-1) | Older European CSVs | `encoding="latin-1"` |
| **cp1252** (Windows-1252) | Microsoft Excel / Windows | `encoding="cp1252"` |
| **utf-16** | Excel exports, Windows text files | `encoding="utf-16"` |
| **utf-16-le** | UTF-16 Little Endian | `encoding="utf-16-le"` |
| **utf-16-be** | UTF-16 Big Endian | `encoding="utf-16-be"` |
| **ascii** | Plain English text only | `encoding="ascii"` |
| **utf-32** | Rare | `encoding="utf-32"` |
| **mac_roman** | Very old Mac files | `encoding="mac_roman"` |

## Quick Rule

1. Try the default:
```python
pd.read_csv("file.csv")
```

2. If you get `UnicodeDecodeError`:
```python
pd.read_csv("file.csv", encoding="latin-1")
```

3. If it still doesn't work:
```python
pd.read_csv("file.csv", encoding="cp1252")
```

4. If it came from Excel:
```python
pd.read_csv("file.csv", encoding="utf-16")
```

## Encodings to Remember

- ✅ `utf-8` → Default, modern standard
- ✅ `latin-1` → Older European CSVs
- ✅ `cp1252` → Windows / Excel
- ✅ `utf-16` → Unicode Excel exports

# 14. Converters

````python
def rename(name):
    if name == "Tottenham Hotspur":
        return "TOT"

    elif name == "Dortmund":
        return "BVB"

    else:
        return name
````

````python
pd.read_csv(
    "/kaggle/input/datasets/johntocci/champions-league-matches-2025-2026/champions_league_matches.csv",
    converters={
        "home_team" : rename,
        "away_team" : rename
    }
)
````
