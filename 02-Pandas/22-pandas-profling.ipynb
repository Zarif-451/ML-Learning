# How to Analyze a `ydata-profiling` Report (Titanic Dataset)

> **Goal:**  
> Do **not** simply scroll through the HTML report. Instead, analyze it like a **Data Scientist** by asking questions and making preprocessing decisions.

---

# Step 1 — Dataset Overview

The first section provides a quick summary of the dataset.

### Questions to Ask

- How many rows are there?
- How many columns are there?
- Is this a **Classification** or **Regression** problem?
- What is the target variable?

### Titanic Example

```text
Rows      : 891
Columns   : 12
Target    : Survived
Problem   : Binary Classification
```

### What should you think?

> I need to build a model that predicts whether a passenger survived or not.

---

# Step 2 — Missing Values

This should be the **first thing** you inspect.

### Questions to Ask

- Which columns contain missing values?
- How many values are missing?
- Is the percentage of missing values acceptable?

### Titanic Example

```text
Cabin

77% Missing
```

Ask yourself:

- Should I drop this column?
- Can I extract useful information before dropping it?
- Should I impute the missing values?

---

Another example:

```text
Age

20% Missing
```

Possible decisions:

- Fill using Mean?
- Fill using Median?
- Predict missing ages using another model?

---

# Step 3 — Identify Variable Types

Classify every feature.

| Feature | Type | Possible Action |
|----------|------|-----------------|
| PassengerId | Identifier | Drop |
| Name | Text | Extract Titles |
| Sex | Categorical | Encode |
| Age | Numerical | Handle Missing Values |
| Fare | Numerical | Check Skewness & Scaling |
| Cabin | Categorical | Handle Missing Values |
| Embarked | Categorical | Handle Missing Values |

### Questions to Ask

- Is this feature useful?
- Does it need preprocessing?
- Should it be dropped?

---

# Step 4 — Warnings Section

This is one of the **most important sections**.

Look carefully for warnings such as:

- Missing Values
- Constant Columns
- High Cardinality
- Highly Correlated Features
- Highly Skewed Features

### Example

```text
Highly Correlated Features
```

Ask yourself:

- Which features?
- Should one of them be removed?

---

Example

```text
Highly Skewed Feature
```

Ask yourself:

- Should I apply a Log Transformation?

---

Example

```text
High Cardinality
```

Ask yourself:

- Which encoding technique should I use?

---

# Step 5 — Analyze Numerical Variables

For every numerical feature, inspect its distribution.

Example:

```text
Age
```

Questions:

- Is it normally distributed?
- Is it skewed?
- Are there outliers?
- Are there missing values?

---

Example:

```text
Fare
```

Questions:

- Is it right-skewed?
- Are there expensive outliers?
- Does it require Log Transformation?
- Does it require Scaling?

---

# Step 6 — Analyze Categorical Variables

Inspect every categorical feature.

Example:

```text
Sex
```

Questions:

- Which category appears most frequently?
- Is the dataset balanced?

---

Example:

```text
Embarked
```

Questions:

- Which port is most common?
- Are there rare categories?
- Are there missing values?

---

Example:

```text
Pclass
```

Questions:

- Is one passenger class much larger than the others?
- Is the dataset imbalanced?

---

# Step 7 — Correlation Analysis

Now inspect relationships between numerical variables.

Example:

```text
Age vs Fare

Correlation = 0.10
```

Interpretation:

- Weak relationship
- Probably not very useful

---

Example:

```text
Fare vs Pclass

Correlation = -0.60
```

Interpretation:

- Strong negative relationship
- Higher-class passengers generally paid higher fares.

Now ask:

- Does this relationship make sense?
- Should I investigate further?

---

# Step 8 — Feature Interactions

Study how features interact with one another.

Examples:

```text
Age vs Survived
```

Question:

- Do children survive more often?

---

```text
Sex vs Survived
```

Question:

- Did women survive more frequently?

---

```text
Pclass vs Fare
```

Question:

- Does passenger class explain ticket price?

---

# Step 9 — Duplicate Rows

Check whether duplicate records exist.

Questions:

- Are there duplicate rows?
- Should duplicates be removed?

---

# Step 10 — Summarize Your Findings

After finishing the report, **stop reading**.

Instead, write your observations.

Example:

```text
1. Cabin contains too many missing values.

2. Age has missing values.

3. Fare is highly right-skewed.

4. PassengerId is only an identifier.

5. Name contains titles that may be useful.

6. Sex must be encoded.

7. Embarked contains missing values.

8. Fare contains outliers.

9. Fare and Pclass are strongly related.
```

This summary becomes your **preprocessing plan**.
