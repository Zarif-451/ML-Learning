# Introduction

**Seaborn** is a Python data visualization library built on top of **Matplotlib**. It provides a high-level interface for creating attractive, informative, and statistically meaningful visualizations with minimal code.

Seaborn integrates seamlessly with **Pandas DataFrames** and is widely used during **Exploratory Data Analysis (EDA)** to understand data distributions, relationships between variables, and categorical comparisons.

**Objective:** Learn the essential Seaborn visualizations required for EDA and Machine Learning while using the Titanic dataset.

# Count Plot
## A Count Plot counts the number of observations in each category and displays them as bars.

## Count Plot vs Bar Chart

| Feature | `sns.countplot()` | `plt.bar()` |
|----------|-------------------|-------------|
| Counting | Automatically counts categories | You provide the counts manually |
| Input | Works directly with a Pandas DataFrame | Requires x and y values |
| Best Use | Displaying frequencies of categorical variables | Creating general-purpose bar charts |
| Code Length | Shorter and simpler | More code required |
| Common Usage | Exploratory Data Analysis (EDA) | General data visualization |

````python
sns.countplot(data=df, x="Sex")
plt.title("Passenger count by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.grid(axis="y")
````

````python
sns.countplot(data=df, x="Survived")
plt.title("Graph of Survival in Titanic")
plt.xlabel("Survived = 1, Didn't Survive = 0")
plt.grid(axis="y")
````

# Histogram (`sns.histplot()`)

A **Histogram** is used to visualize the **distribution of a numerical variable** by dividing the data into intervals (bins) and counting how many observations fall into each interval.

In **Exploratory Data Analysis (EDA)**, histograms help identify the shape of the data, detect skewness, locate outliers, and understand how values are distributed before applying Machine Learning algorithms.

## Recommended Number of Bins

| Dataset Size | Recommended Number of Bins |
|--------------|---------------------------:|
| < 100 rows | 5–10 |
| 100–1,000 rows | **10–20 ⭐** |
| 1,000–10,000 rows | 20–40 |
| > 10,000 rows | 30–60 |

````python
df.shape[0]
````

````python
sns.histplot(data=df, x="Age", bins=15)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
````

````python
sns.histplot(data=df, x="Fare", bins=15)
plt.title("Fare Distribution")
````

# Kernel Density Estimation (KDE)

A **Kernel Density Estimation (KDE)** plot is a smooth curve that estimates the probability distribution of a numerical variable.

Unlike a histogram, which groups data into discrete bins, a KDE plot provides a continuous estimate of the underlying distribution. It is commonly used during **Exploratory Data Analysis (EDA)** to identify the shape of the data, detect skewness, and compare distributions.

## Histogram vs KDE

| Histogram | KDE |
|-----------|-----|
| Uses bars | Uses a smooth curve |
| Depends on the number of bins | Does not require bins |
| Shows the frequency (count) of observations | Shows the estimated probability density |
| Simple and intuitive for beginners | Better for understanding the overall shape of the distribution |
| Can change noticeably when the number of bins changes | Less affected by arbitrary grouping choices |
| Best for visualizing counts | Best for identifying the underlying distribution pattern |

````python
sns.kdeplot(data=df, x="Age")
````

````python
sns.histplot(data=df, x="Age", bins=20)

plt.show()
````

````python
sns.histplot(data=df, x="Age", bins=20, kde=True)
````

### Observations

- Most passengers were between **20 and 35 years old**.
- The highest density occurs around **25 years of age**.
- The age distribution is **right-skewed (positively skewed)**.
- There are relatively few elderly passengers above **60 years**.
- The KDE curve closely follows the overall shape of the histogram while providing a smoother estimate of the distribution.

# Box Plot (`sns.boxplot()`)

A **Box Plot** is a statistical visualization used to summarize the distribution of a numerical variable. It displays the median, quartiles, spread of the data, and potential outliers.

In **Exploratory Data Analysis (EDA)**, box plots are widely used to detect outliers, compare distributions across different categories, and understand the variability of numerical features before applying Machine Learning algorithms.

## Anatomy of a Box Plot

```text
                 ●      ●
                  ↑ Outliers

───────────────┬────────────────
      ┌───────────────────┐
      │                   │
──────┤─────────│─────────├──────
     Q1      Median      Q3
```

### Components

- **Left edge of the box** → **Q1 (25th Percentile)**
- **Line inside the box** → **Median (50th Percentile)**
- **Right edge of the box** → **Q3 (75th Percentile)**
- **Lines extending from the box** → **Whiskers**
- **Points outside the whiskers** → **Potential Outliers**

````python
sns.boxplot(data=df, x="Age")
plt.title("Age Distribution")
````

````python
sns.boxplot(data=df, x="Fare")
````

In **Exploratory Data Analysis (EDA)**, violin plots are useful for comparing the distribution of a numerical feature across different categories while also revealing the shape of the distribution.

## Interpreting Correlation Coefficients

The correlation coefficient ranges from **-1** to **+1**.

| Correlation Coefficient ($r$) | Strength | Interpretation |
|:-----------------------------:|----------|----------------|
| $+1.00$ | Perfect Positive | As one variable increases, the other always increases. |
| $+0.70$ to $+0.99$ | Strong Positive | Strong positive relationship. |
| $+0.30$ to $+0.69$ | Moderate Positive | Moderate positive relationship. |
| $+0.01$ to $+0.29$ | Weak Positive | Weak positive relationship. |
| $0.00$ | No Correlation | No linear relationship. |
| $-0.01$ to $-0.29$ | Weak Negative | Weak negative relationship. |
| $-0.30$ to $-0.69$ | Moderate Negative | Moderate negative relationship. |
| $-0.70$ to $-0.99$ | Strong Negative | Strong negative relationship. |
| $-1.00$ | Perfect Negative | As one variable increases, the other always decreases. |

> **Note:**
>
> - The **sign (+/-)** indicates the **direction** of the relationship.
>   - **Positive (+):** Both variables move in the same direction.
>   - **Negative (-):** Variables move in opposite directions.
>
> - The **absolute value** (distance from 0) indicates the **strength** of the relationship.
>   - Closer to **1** → Strong relationship.
>   - Closer to **0** → Weak relationship.
>
> **Correlation does not imply causation.** A high correlation means two variables tend to change together, but it does **not** prove that one variable causes the other.

### Quick Rule of Thumb

- **0.00 – 0.29** → Weak
- **0.30 – 0.69** → Moderate
- **0.70 – 1.00** → Strong

The **sign (+/-)** tells the **direction**:
- **+** → Move together
- **−** → Move in opposite directions

## Pair Plot (`sns.pairplot()`)

A **Pair Plot** is a visualization that displays the pairwise relationships between multiple numerical variables in a dataset. It automatically generates **scatter plots** for every pair of numerical features and **histograms** (or KDE plots) along the diagonal to show the distribution of each feature.

In **Machine Learning** and **Exploratory Data Analysis (EDA)**, pair plots are useful for identifying relationships, trends, clusters, distributions, and outliers among multiple variables simultaneously.

### Mathematical Representation

For a dataset with numerical features

$$
X = \{X_1, X_2, X_3, \ldots, X_n\},
$$

a pair plot visualizes every pairwise combination

$$
(X_i, X_j), \quad \forall\, i,j \in \{1,2,\ldots,n\}.
$$

The total number of scatter plots generated is

$$
\frac{n(n-1)}{2},
$$

while the diagonal contains the distribution of each feature,

$$
(X_1, X_1),\ (X_2, X_2),\ \ldots,\ (X_n, X_n).
$$

Thus, a pair plot provides a comprehensive overview of the relationships and distributions of all numerical variables in a single figure.

````python
sns.pairplot(
    df[
        [
            "Age",
            "Fare",
            "SibSp",
            "Parch"
        ]
    ]
)
````
