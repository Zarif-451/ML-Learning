# Introduction

**Matplotlib** is a Python library used for creating graphs and charts to visualize data. It helps transform raw numerical data into meaningful visual representations, making it easier to identify patterns, trends, distributions, and relationships.

In Machine Learning, Matplotlib is widely used during **Exploratory Data Analysis (EDA)** to understand datasets before building models. It serves as the foundation for many visualization libraries, including **Seaborn**.

**Objective:** Learn the fundamental plotting techniques in Matplotlib that are essential for data visualization and future Machine Learning workflows.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/kaggle/input/datasets/shuofxz/titanic-machine-learning-from-disaster/train.csv")
```

# Making the Plot Readable

```python
plt.plot(df["Fare"].head(10))
plt.title("Fare of the first 10 Passengers in Titanic")
plt.xlabel("Passenger Index")
plt.ylabel("Fare")
plt.grid()
```

```python
plt.plot(df["Age"].head(10))
plt.title("Age of first 10 Passengers")
plt.xlabel("Passenger Index")
plt.ylabel("Age")
plt.grid()
```

# Changing the Figure Size (figsize) 

```python
plt.figure(figsize=(8, 5))  #plt.figure(figsize=(width, height))
plt.plot(df["Fare"].head(10))
plt.title("Fare of First 10 Passengers")
plt.xlabel("Passenger Index")
plt.ylabel("Fare")
plt.grid()
```

# Colors, Line Styles and Markers

```python
plt.plot(df["Fare"].head(10), color="red", linestyle="--", marker="s")
plt.title("Fare of First 10 Passengers")
plt.xlabel("Passenger Index")
plt.ylabel("Fare")
plt.grid()
```

```python
plt.plot(df["Fare"].head(10), "ro-")
```

# Plotting Multiple Lines

```python
plt.plot(df["Fare"].head(10), label="Fare")
plt.plot(df["Survived"].head(10), label="Survived")
plt.legend()  #Shows which line represents which
plt.grid()
```

# Bar Chart

A **Bar Chart** is a graphical representation used to compare **categorical data**. It displays categories along one axis and their corresponding values or frequencies as rectangular bars, where the height (or length) of each bar is proportional to the represented value.

Bar charts are widely used in **Exploratory Data Analysis (EDA)** to visualize and compare the frequency or count of different categories, making it easier to identify patterns, differences, and distributions within a dataset.

**Examples:** Number of male vs. female passengers, passenger class distribution, and survival counts in the Titanic dataset.

| **Line Plot (`plt.plot()`)** | **Bar Chart (`plt.bar()`)** |
|:----------------------------:|:---------------------------:|
| Continuous data | Categorical data |
| Shows trends over time | Compares categories |
| Connects data points with lines | Represents values using separate bars |

```python
gender_counts = df["Sex"].value_counts()
```

```python
plt.bar(gender_counts.index, gender_counts.values)
plt.grid(axis='y')
print(gender_counts)
```

```python
class_counts = df["Pclass"].value_counts()
plt.bar(class_counts.index, class_counts.values)
plt.title("Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.grid(axis="y")
print(class_counts)
```

```python
survive_counts = df["Survived"].value_counts()
plt.bar(survive_counts.index, survive_counts.values)
print(survive_counts)
```

# Scatter Plot
## Numerical Variable Vs Numerical Variable

```python
plt.scatter(df["Age"], df["Fare"])
plt.title("Age Vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.grid()
```

# Pie Chart 
## Vs. Comparison

```python
gender_counts = df["Sex"].value_counts()
print(gender_counts)

plt.pie(gender_counts.values, labels=gender_counts.index, autopct="%1.1f%%")
plt.title("Passenger Gender Distribution")
```

```python
survive_counts = df["Survived"].value_counts()
print(survive_counts)

plt.pie(survive_counts.values, 
        labels=["Did not Survive", "Survived"],
        autopct="%1.1f%%")
```
